```python
def count_vowels (text):
    count=0
    for ch in text :
        if ch in "aeiouAEIOU":
            count += 1
    return count
print(count_vowels("welcome"))
```

Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested functions:

```python
def calculate_rectangle(length, width):

    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    area = calculate_area()
    perimeter = calculate_perimeter()

    return area, perimeter


area, perimeter = calculate_rectangle(5, 3)

print("Area:", area)
print("Perimeter:", perimeter)
```

. Write a nested function to calculate the total hotel bill based on the number of nights. A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

```python
def hotel_bill(nights):

    def room_cost():
        return nights * 5000

    def calculate_tax():
        return room_cost() * 0.10

    cost = room_cost()
    tax = calculate_tax()
    total = cost + tax

    print("Room Cost:", cost)
    print("Tax:", tax)
    print("Total Bill:", total)
hotel_bill(3)
```

Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

An inner function calculate_total() → multiplies price * quantity.
An inner function apply_discount(total) → subtracts discount from total (but never below 0).

```python
def shopping_bill(price, quantity, discount=0):

    def calculate_total():
        return price * quantity

    def apply_discount(total):
        return max(0, total - discount)

    total = calculate_total()
    final_bill = apply_discount(total)

    print("Total:", total)
    print("Discount:", discount)
    print("Final Bill:", final_bill)


shopping_bill(1000, 3, 500)
```
