Make a class called Employee. The init() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri- butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each instance. Then do some updations with it's methods.

````python

class Employee:

    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):

        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    # Method to display employee details
    def describe_employee(self):
        print("\n----- Employee Details -----")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Experience:", self.year_of_experience, "years")

    # Method to update experience
    def update_year_of_experience(self, years):
        self.year_of_experience += years
        print("Updated Experience:", self.year_of_experience)

    # Method to increase salary
    def increase_salary(self, amount):
        self.salary += amount
        print("Updated Salary:", self.salary)


# Creating first employee
employee101 = Employee(
    "Ali", 101, "Karachi", 50000, "Manager",
    2020, "Masters", 5
)

# Printing four attributes individually
print(employee101.name)
print(employee101.number)
print(employee101.address)
print(employee101.salary)

# Calling all methods
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(1)


# Creating 10 different employee instances
employee102 = Employee("Ahmed", 102, "Karachi", 40000, "Teacher", 2021, "Masters", 4)
employee103 = Employee("Sara", 103, "Lahore", 45000, "Accountant", 2019, "B.Com", 6)
employee104 = Employee("Ayesha", 104, "Karachi", 55000, "Principal", 2018, "M.Ed", 8)
employee105 = Employee("Usman", 105, "Islamabad", 35000, "Clerk", 2022, "B.A", 3)
employee106 = Employee("Fatima", 106, "Karachi", 60000, "Coordinator", 2017, "M.A", 9)
employee107 = Employee("Hassan", 107, "Hyderabad", 30000, "Assistant", 2023, "B.Com", 2)
employee108 = Employee("Zainab", 108, "Lahore", 42000, "Teacher", 2020, "B.Ed", 5)
employee109 = Employee("Bilal", 109, "Karachi", 38000, "Cashier", 2021, "B.A", 4)
employee110 = Employee("Maryam", 110, "Islamabad", 48000, "HR Manager", 2019, "MBA", 6)

# Store all employees in a list
employees = [
    employee101, employee102, employee103,
    employee104, employee105, employee106,
    employee107, employee108, employee109,
    employee110
]

# Display details of all 10 employees
for employee in employees:
    employee.describe_employee()

# Updating employees
employee102.increase_salary(3000)
employee103.update_year_of_experience(2)
employee104.increase_salary(5000)
employee105.update_year_of_experience(1)
````

````python

# Creating employee details using a function

def create_employee(name, number, address, salary, post,
                    year_of_joining, education, experience):

    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "experience": experience
    }


# Function to describe employee

def describe_employee(employee):
    print("\n----- Employee Details -----")
    for key, value in employee.items():
        print(key, ":", value)


# Function to update experience

def update_year_of_experience(employee, years):
    employee["experience"] += years
    print("Updated Experience:", employee["experience"])


# Function to increase salary

def increase_salary(employee, amount):
    employee["salary"] += amount
    print("Updated Salary:", employee["salary"])


# Creating first employee

employee101 = create_employee(
    "Ali", 101, "Karachi", 50000,
    "Manager", 2020, "Masters", 5
)

# Printing four attributes individually

print(employee101["name"])
print(employee101["number"])
print(employee101["address"])
print(employee101["salary"])

# Calling all functions

describe_employee(employee101)
increase_salary(employee101, 5000)
update_year_of_experience(employee101, 1)


# Creating 9 more employees

employee102 = create_employee("Ahmed", 102, "Karachi", 40000, "Teacher", 2021, "Masters", 4)

employee103 = create_employee("Sara", 103, "Lahore", 45000, "Accountant", 2019, "B.Com", 6)

employee104 = create_employee("Ayesha", 104, "Karachi", 55000, "Principal", 2018, "M.Ed", 8)

employee105 = create_employee("Usman", 105, "Islamabad", 35000, "Clerk", 2022, "B.A", 3)

employee106 = create_employee("Fatima", 106, "Karachi", 60000, "Coordinator", 2017, "M.A", 9)

employee107 = create_employee("Hassan", 107, "Hyderabad", 30000, "Assistant", 2023, "B.Com", 2)

employee108 = create_employee("Zainab", 108, "Lahore", 42000, "Teacher", 2020, "B.Ed", 5)

employee109 = create_employee("Bilal", 109, "Karachi", 38000, "Cashier", 2021, "B.A", 4)

employee110 = create_employee("Maryam", 110, "Islamabad", 48000, "HR Manager", 2019, "MBA", 6)


# Store all employees in a list

employees = [
    employee101, employee102, employee103,
    employee104, employee105, employee106,
    employee107, employee108, employee109,
    employee110
]


# Display details of all employees

for employee in employees:
    describe_employee(employee)


# Updating employee records

increase_salary(employee102, 3000)
update_year_of_experience(employee103, 2)

increase_salary(employee104, 5000)
update_year_of_experience(employee105, 1)
````
