Objective

Python Project: Online Shopping Cart

Create a simple shopping cart program where a customer can view products, add items to a cart, view the bill and
complete checkout.
Products
Product ID Product Price Available Stock
P1 Shoes Rs. 2,500 5
P2 Shirt Rs. 1,500 10
P3 Bag Rs. 3,000 4
P4 Watch Rs. 4,000 3
P5 Headphones Rs. 2,000 6
Program Requirements
1. View Products: Display all five products with product ID, name, price and available stock. Show them in a clear,
readable format (not as one raw dictionary).
2. Add to Cart: Ask for a product ID and quantity. Add the item only when the product exists and enough stock is
available.
3. View Cart: Display each selected product, quantity, item subtotal and the cart total.
4. Checkout: Calculate the final bill. Give a 10% discount when the cart total is Rs. 5,000 or more. After checkout,
reduce product stock and empty the cart.
5. Exit: Allow the customer to close the program.
Python Concepts to Use
Dictionaries and nested dictionaries, while loop, for loop, if-elif-else conditions, input(), calculations and print().
Important Notes
• Use a menu that keeps running until the customer selects Exit.
• Do not allow the customer to add more items than the available stock.
• When the same product is added again, combine its quantity in the cart and check the total requested quantity against stock.
• After a successful checkout, clear the cart so a new order can be started.

```python

# Online Shopping Cart Project

# Product Dictionary
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

# Empty Cart
cart = {}

# Main Menu
while True:

    print("\n===== ONLINE SHOPPING CART =====")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice: ")

    # View Products
    if choice == "1":

        print("\n===== AVAILABLE PRODUCTS =====")

        for product_id, details in products.items():
            print("--------------------------------")
            print("Product ID:", product_id)
            print("Product Name:", details["name"])
            print("Price: Rs.", details["price"])
            print("Available Stock:", details["stock"])

    # Add to Cart
    elif choice == "2":

        product_id = input("Enter Product ID: ").upper()

        if product_id in products:

            quantity = int(input("Enter quantity: "))

            if quantity > 0:

                cart_quantity = cart.get(product_id, 0)

                total_quantity = cart_quantity + quantity

                if total_quantity <= products[product_id]["stock"]:

                    cart[product_id] = total_quantity

                    print("Product added to cart successfully!")

                else:
                    print("Sorry! Not enough stock available.")

            else:
                print("Please enter a valid quantity.")

        else:
            print("Invalid Product ID!")

    # View Cart
    elif choice == "3":

        if len(cart) == 0:

            print("Your cart is empty.")

        else:

            print("\n===== YOUR SHOPPING CART =====")

            cart_total = 0

            for product_id, quantity in cart.items():

                name = products[product_id]["name"]
                price = products[product_id]["price"]

                subtotal = price * quantity

                cart_total += subtotal

                print("--------------------------------")
                print("Product:", name)
                print("Quantity:", quantity)
                print("Subtotal: Rs.", subtotal)

            print("--------------------------------")
            print("Cart Total: Rs.", cart_total)

    # Checkout
    elif choice == "4":

        if len(cart) == 0:

            print("Your cart is empty. Add products first.")

        else:

            total = 0

            print("\n===== FINAL BILL =====")

            for product_id, quantity in cart.items():

                name = products[product_id]["name"]
                price = products[product_id]["price"]

                subtotal = price * quantity

                total += subtotal

                print(name, "x", quantity, "= Rs.", subtotal)

            print("--------------------------------")
            print("Original Total: Rs.", total)

            if total >= 5000:

                discount = total * 0.10

            else:

                discount = 0

            final_bill = total - discount

            print("Discount: Rs.", discount)
            print("Final Bill: Rs.", final_bill)

            # Reduce Product Stock
            for product_id, quantity in cart.items():

                products[product_id]["stock"] -= quantity

            # Empty Cart
            cart.clear()

            print("\nCheckout Successful!")
            print("Thank you for shopping!")

    # Exit
    elif choice == "5":

        print("Thank you! Visit again.")
        break

    else:

        print("Invalid choice! Please select 1 to 5.")
```
